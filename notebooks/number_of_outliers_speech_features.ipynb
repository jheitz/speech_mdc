```python
feature_name_match = {}

manual_mapping = {
    'lit_VP -> VBG': 'VP -\\textgreater VBG',
    'lit_VP -> VBG_PP': 'VP -\\textgreater VBG\\_PP',
    'lit_VP -> VBD_NP': 'VP -\\textgreater VBD\\_NP',
    'lit_INTJ -> UH': 'INTJ -\\textgreater UH',
    'lit_mattr': 'moving\\_average\\_type\\_token\\_ratio',
    'lit_NDW-50': 'n\\_unique\\_in\\_50',
    'sung_AoA_noun': 'age\\_of\\_acquisition\\_noun',
    'sung_concreteness_noun': 'noun\\_concreteness',
    'lit_subordinate_coordinate_conjunction_ratio': 'subordinate\\_coordinate\\_con-junction\\_ratio',
    'lit_NP -> PRP': 'NP -\\textgreater PRP',
}

def clean_feature(feature):
    try:
        return feature.replace("\\", "").replace("_", "").replace(" ", "").replace("lit","").replace("sung","").replace("pause","").replace("smile","").replace("phon","").lower()
    except:
        print(f"Attention: Cannot clean {feature}")
        return None

for f in outlier_counts.Feature.tolist():
    if pd.isna(f):
        continue
    elif f in manual_mapping:
        feature_name_match[f] = manual_mapping[f]
    elif clean_feature(f) in [clean_feature(name) for name in feature_names_latex]:
        matches = [name for name in feature_names_latex if clean_feature(name) == clean_feature(f)]
        if len(matches) > 1:
            print(f"Attention: Multiple matches for {f}: {matches}")
        feature_name_match[f] = matches[0]
    else:
        print(f"No match for {f}")
```
