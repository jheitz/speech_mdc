```python
import pandas as pd

acs_outliers_removed_2024 = pd.read_csv("/Users/jheitz/git/luha-prolific-study/data/processed_combined/data/acs_outcomes_outliers_removed.csv")
acs_outliers_removed_2026 = pd.read_csv("/Users/jheitz/git/luha-prolific-study/data2026/processed_combined/data/acs_outcomes_outliers_removed.csv")

relevant_tests = {
    'connect_the_dots_I_time_msec': 'Trail Making Test A',
    'connect_the_dots_II_time_msec': 'Trail Making Test B',
    'wordlist_correct_words': 'RAVLT (Learning)',
    'wordlist_delayed_correct_words': 'RAVLT (Recall)',
    'wordlist_recognition_correct_words': 'RAVLT (Recognition)',
    'avg_reaction_speed': 'Visual Reaction Time',
    'box_tapping_total_time': 'Corsi Block-tapping Test',
    'fill_the_grid_total_time': 'Grooved Pegboard',
    'digit_sequence_1_correct_series': 'Digit Span (forward)',
    'digit_sequence_2_correct_series': 'Digit Span (backward)',
    'clickskill_time': 'Clicking speed test',
    'dragskill_time': 'Mouse dragging speed test'
}
```

```python
cols =list(acs_outliers_removed_2024.columns)
[c for c in cols if 'wordlist' in c]
```

```python
missing_count_2024 = acs_outliers_removed_2024[relevant_tests.keys()].isna().sum()
missing_count_2026 = acs_outliers_removed_2026[relevant_tests.keys()].isna().sum()
total_count_2024 = acs_outliers_removed_2024.shape[0]
total_count_2026 = acs_outliers_removed_2026.shape[0]
stats = pd.DataFrame({'2024': missing_count_2024, '2026': missing_count_2026, 'total_2024': total_count_2024, 'total_2026': total_count_2026})
stats
```

```python
stats_combined = pd.DataFrame({'# Outliers': missing_count_2024 + missing_count_2026, '# Assessments': total_count_2024 + total_count_2026})
stats_combined['Outlier percentage'] = (stats_combined['# Outliers'] / stats_combined['# Assessments'] * 100).apply(lambda x: f"{x:.2f}%")
stats_combined.index.name = "Cognitive Test"
stats_combined = stats_combined.rename(index=relevant_tests)
stats_combined
```

```python
stats_combined = stats_combined[stats_combined.index != 'Overall']
overall_statistic = pd.DataFrame({'# Outliers': [stats_combined['# Outliers'].sum()], '# Assessments': [stats_combined['# Assessments'].sum()]})
overall_statistic['Outlier percentage'] = (overall_statistic['# Outliers'] / overall_statistic['# Assessments'] * 100).apply(lambda x: f"{x:.2f}%")
overall_statistic.index = ['Overall']

stats_combined = pd.concat([stats_combined, overall_statistic])
stats_combined
```

```python
from datetime import datetime
now = lambda: datetime.now().strftime("%Y-%m-%d %H:%M")
print("%", now())
print(stats_combined.to_latex(index=True, caption="Number and percentage of individual ACS test scores marked as outliers and subsequently imputed, across all 1481 assessments.", label="s_tab:outliers_cognitive_tests", float_format="%.0f").replace("%", "\\%").replace("#", "\\#"))
```
