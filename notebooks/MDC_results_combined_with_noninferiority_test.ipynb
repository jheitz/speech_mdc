# Combined results tables for the four targets and both cohorts

Reads the four per-target CSVs produced by `MDC_calculation_and_SEM_analysis.ipynb`
(`MDC_results/{target}.csv` for `language`, `memory`, `executive_function`, `speed`)
and combines them into latex tables per cohort.

```python
import json

import pandas as pd
from pandas import MultiIndex, IndexSlice
from datetime import datetime
now = lambda: datetime.now().strftime("%Y-%m-%d %H:%M")

all_targets = ['language', 'memory', 'executive_function', 'speed']

target_renaming = {
    'language': 'Language',
    'executive_function': 'Executive Function',
    'speed': 'Speed',
    'memory': 'Memory'
}
task_renaming = {
    'cookieTheft': 'Cookie Theft',
    'picnicScene': 'Picnic Scene',
    'journaling': 'Journaling',
    'avg_speech_prediction': 'Avg. Speech Prediction',
}
```

```python
dfs = {}
for target in all_targets:
    df = pd.read_csv(f'MDC_results/{target}.csv', index_col=0)
    df.columns = [f"Cohort {c}" for c in df.columns]
    df.index = [idx.replace("_"+target, "") for idx in df.index]
    dfs[target] = df

combined = pd.concat(dfs, axis=1).sort_index(axis=1, level=1)
combined
```

```python

tasks = ['cookieTheft', 'picnicScene', 'journaling']
cohorts = ['Cohort A', 'Cohort B']
combined_tasks = []

def combine_metrics(series):
    display(series)
    task = list(series.index)[0].split("_")[0]
    aggregated = pd.Series({task: "/".join(series.round(2).astype(str))}, name=series.name)
    return aggregated


for cohort in cohorts:
    for task in tasks:
        combined_task_separate = combined.loc[[
            f"{task}_sem_tli",
            f"{task}_sem_cfi",
            f"{task}_sem_srmr",
            #f"{task}_sem_chi2",
        ], pd.IndexSlice[:, cohort]].astype(float)
        combined_task_separate.columns = combined_task_separate.columns.droplevel(1)
        combined_task = combined_task_separate.apply(combine_metrics, axis=0)
        combined_tasks.append(combined_task)

    combined_subset = pd.concat(combined_tasks, axis=0)[all_targets].rename(columns=target_renaming)
    combined_subset.index = [task_renaming.get(c,c) for c in combined_subset.index]
    display(combined_subset)

    break
```

```python
tasks = ['cookieTheft', 'picnicScene', 'journaling']
cohorts = ['Cohort A', 'Cohort B']
metric_keys = ['sem_tli', 'sem_cfi', 'sem_srmr']
metric_names = ['TLI', 'CFI', 'SRMR']

for cohort in cohorts:
    combined_tasks = []

    for task in tasks:
        row = combined.loc[
            [f"{task}_{m}" for m in metric_keys],
            pd.IndexSlice[all_targets, cohort]
        ].astype(float)
        row.columns = row.columns.droplevel(1)
        row.index = metric_names
        # row: (metric x target) -> transpose to (target x metric)
        row = row.T
        # build one row with (target, metric) MultiIndex columns
        tuples = [(target, metric) for target in row.index for metric in metric_names]
        values = row.values.flatten()
        combined_tasks.append(
            pd.DataFrame(
                [values],
                index=[task_renaming.get(task, task)],
                columns=pd.MultiIndex.from_tuples(tuples)
            )
        )

    combined_subset = pd.concat(combined_tasks)
    combined_subset.columns = pd.MultiIndex.from_tuples(
        [(target_renaming.get(t, t), m) for t, m in combined_subset.columns],
        names=[None, None]
    )

    combined_subset_latex = combined_subset.to_latex(float_format="%.2f", multirow=True, multicolumn=True, multicolumn_format="c", column_format="l" + "c" * len(combined_subset.columns),
                                                     caption=f"Absolute fit indices of the speech measurement model for each speech task and composite score for {cohort}. CFI and TLI above 0.95 and SRMR below 0.08 indicate adequate fit. Fit indices for the five alternative models and the corresponding likelihood-ratio tests are reported in the supplementary material (Tables~\\ref{{tab:sem_fit_cohortA}} and \\ref{{tab:sem_lr_cohortA}}).")
    print(f"% === {cohort}: full-model summary {cohort} ({now()}) ===")
    print(combined_subset_latex)
```

```python
from io import StringIO
import json


def format_num(x, decimals=3):
    if pd.isna(x):
        return "---"
    return f"{x:.{decimals}f}"

def format_p(p):
    """Format p-values: '<0.001' for very small, else 3 decimals."""
    if pd.isna(p):
        return "---"
    if p < 0.001:
        return "$<0.001$"
    return f"{p:.3f}"


task_display_names = {
    "cookieTheft":                 r"Cookie Theft",
    "picnicScene":                 r"Picnic Scene",
    "journaling":                  r"Journaling",
    "avg_speech_prediction":       r"Avg.\ (all tasks)",
    "avg_cookieTheft_picnicScene": r"Avg.\ (Cookie Theft + Picnic Scene)",
}

def combined_lr_table_to_latex(
    results_df,
    cohort='Cohort A',
    targets=('language', 'memory', 'executive_function', 'speed'),
    label='tab:sem_lr_cohortA_all_targets',
    caption=None,
):
    """
    Create a wide LR-test LaTeX table with one 3-column block per target:
        Δχ² | Δdf | p

    Rows:
        task × comparison
    """

    comparison_display = {
        'no_attenuation vs full':         r'$\lambda=1$ vs.\ Full',
        'no_beta vs full':                r'No $\beta$ vs.\ Full',
        'no_attenuation_no_beta vs full': r'$\lambda=1$, no $\beta$ vs.\ Full',
        'full vs free_hat_resid_by_wave': r'Full vs.\ free $\sigma_\epsilon^2$ by wave',
        'full vs free_y_resid_by_wave':   r'Full vs.\ free $\sigma_\delta^2$ by wave',
    }

    # ------------------------------------------------------------
    # Parse all target LR dfs
    # ------------------------------------------------------------
    lr_by_target = {}

    for target in all_targets:

        # json-encoded dataframe string
        #print((target, cohort), 'cookieTheft_sem_lr_full')
        #print(results_df.loc[:, (target, cohort)].index.tolist())
        raw = results_df.loc['cookieTheft_sem_lr_full', (target, cohort)]

        # if your actual column is different, replace above with:
        # raw = results_df.loc[(target, cohort), 'sem_lr_json']

        lr_df_text = json.loads(raw)
        lr_df = pd.DataFrame(
            data=lr_df_text["data"],
            columns=lr_df_text["columns"]
        )
        lr_df.index = pd.MultiIndex.from_tuples(
            lr_df_text["index"],
            names=["Task", "Comparison"]
        )

        lr_by_target[target] = lr_df.copy()

    # ------------------------------------------------------------
    # Build unified row index from first target
    # ------------------------------------------------------------
    base_index = lr_by_target[targets[0]].index

    out = pd.DataFrame(index=base_index)

    # ------------------------------------------------------------
    # Add formatted columns for each target
    # ------------------------------------------------------------
    for target in targets:

        df = lr_by_target[target]

        out[(target, r'$\Delta\chi^2$')] = [
            format_num(v, 3)
            for v in df['delta_chi2']
        ]

        out[(target, r'$\Delta$df')] = [
            '---' if pd.isna(v) else f'{int(v)}'
            for v in df['delta_df']
        ]

        out[(target, r'$p$')] = [
            format_p(v)
            for v in df['p_value']
        ]

    # ------------------------------------------------------------
    # Pretty row labels
    # ------------------------------------------------------------
    out.index = pd.MultiIndex.from_tuples(
        [
            (
                task_display_names.get(task, task),
                comparison_display.get(comp, comp),
            )
            for (task, comp) in out.index
        ],
        names=['Task', 'Comparison'],
    )

    # ------------------------------------------------------------
    # Pretty column labels
    # ------------------------------------------------------------
    target_display = {
        'language': r'Language',
        'memory': r'Memory',
        'executive_function': r'Executive function',
        'speed': r'Speed',
    }

    out.columns = pd.MultiIndex.from_tuples([
        (
            target_display.get(target, target),
            stat,
        )
        for (target, stat) in out.columns
    ])

    # ------------------------------------------------------------
    # Column format
    # ll = row index cols
    # then 3 numeric cols per target
    # ------------------------------------------------------------
    column_format = 'll' + ('rrr' * len(targets))

    # ------------------------------------------------------------
    # Caption
    # ------------------------------------------------------------
    if caption is None:
        caption = (
            f"{cohort}: likelihood-ratio tests for the nested model comparisons across all cognitive target scores. For the three nested alternatives (top three rows per task), a significant $p$-value rejects the alternative in favor of the full model. For the two stationarity relaxations (bottom two rows per task), a non-significant $p$-value is consistent with stationary residual variances."
        )

    # ------------------------------------------------------------
    # Export LaTeX
    # ------------------------------------------------------------
    latex = out.to_latex(
        escape=False,
        multicolumn=True,
        multirow=True,
        column_format=column_format,
        caption=caption,
        label=label,
    )

    latex = latex.replace(r'\cline', r'\cmidrule')

    return latex


latex = combined_lr_table_to_latex(
    results_df=combined,
    cohort='Cohort A',
    label='tab:sem_lr_cohortA',
)

print(f"% === Cohort A: LR tests {now()} ===")
print(latex)
```

```python
import pandas as pd

TASK_ORDER = ['journaling', 'cookieTheft', 'picnicScene', 'avg_speech_prediction']
TASK_DISPLAY = {
    'cookieTheft': 'Cookie Theft',
    'picnicScene': 'Picnic Scene',
    'journaling': 'Journaling',
    'avg_speech_prediction': 'Avg. Speech Pred.',
}
DOMAIN_ORDER = ['language', 'memory', 'executive_function',  'speed']
DOMAIN_DISPLAY = {
    'executive_function': 'Executive Function',
    'language': 'Language',
    'memory': 'Memory',
    'speed': 'Speed',
}

N_BOOTSTRAP_SAMPLES = 10000


def _melt_combined(combined):
    """Reshape `combined` from
        index = '<task>_bootstrap_<metric>', columns = (domain, cohort)
    to a long dataframe with columns
        cohort, domain, task, mean_diff, ci_low, ci_high, excludes_zero, prop_speech_worse.
    """
    # Split the row index into (task, metric).
    idx = combined.index.to_series().str.extract(r'^(?P<task>.+)_bootstrap_(?P<metric>.+)$')
    tmp = combined.copy()
    tmp.index = pd.MultiIndex.from_frame(idx)

    # Long form: one row per (task, metric, domain, cohort).
    long = tmp.stack(level=[0, 1], future_stack=True).rename('value').reset_index()
    long.columns = ['task', 'metric', 'domain', 'cohort', 'value']

    # Pivot metrics into columns.
    wide = long.pivot_table(
        index=['cohort', 'domain', 'task'],
        columns='metric',
        values='value',
        aggfunc='first',
    ).reset_index()
    wide.columns.name = None

    # Types.
    wide['excludes_zero'] = wide['excludes_zero'].replace("True", 1).replace("False", 0).astype(bool)
    for c in ['mean_diff', 'ci_low', 'ci_high', 'prop_speech_worse']:
        wide[c] = wide[c].astype(float)
    return wide


def _fmt_ci(row):
    s = f"{row['ci_low']:.2f} -- {row['ci_high']:.2f}"
    if row['excludes_zero']:
        s += '*'
    return s


def _fmt_prop(p):
    return f"{p:.1%}".replace('%', r'\%')


def _build_cohort_wide(long_df, cohort_label):
    """Build the (task x (domain, metric)) wide table for a single cohort."""
    sub = long_df[
        (long_df['cohort'] == f'Cohort {cohort_label}')
        & (long_df['task'].isin(TASK_ORDER))
    ].copy()
    sub['ci_str'] = sub.apply(_fmt_ci, axis=1)
    sub['prop_str'] = sub['prop_speech_worse'].apply(_fmt_prop)

    pieces = []
    for domain in DOMAIN_ORDER:
        d = (
            sub[sub['domain'] == domain]
            .set_index('task')[['ci_str', 'prop_str']]
        )
        d.columns = pd.MultiIndex.from_tuples([
            (DOMAIN_DISPLAY[domain], r'95\% CI'),
            (DOMAIN_DISPLAY[domain], r'\% better' ),
        ])
        pieces.append(d)

    wide = pd.concat(pieces, axis=1).loc[TASK_ORDER]
    wide.index = [TASK_DISPLAY[t] for t in wide.index]
    wide.index.name = None
    return wide


def cohort_to_latex(wide, cohort_label, n_bootstrap=N_BOOTSTRAP_SAMPLES):
    column_format = 'l' + 'c' * wide.shape[1]
    body = wide.to_latex(
        column_format=column_format,
        multicolumn=True,
        multicolumn_format='c',
        escape=False,
    )
    caption = (
        f"Bootstrap test results assessing whether the speech-based MDC is significantly larger than the cognitive assessment's MDC on Cohort {cohort_label}. For each task and composite cognitive score, we report the 95\\% CI of the difference (speech MDC minus cognitive MDC) across {n_bootstrap} bootstrap replicates, and the proportion of replicates where the speech MDC is larger (less sensitive) than the cognitive MDC (\\% larger). An asterisk (*) on the CI indicates that the CI excludes zero, i.e. the speech MDC is significantly larger than the cognitive MDC."
    )
    label = f"tab:mdc_bootstrap_cohort{cohort_label}"
    return (
        "\\begin{table}\n"
        f"\\caption{{{caption}}}\n"
        f"\\label{{{label}}}\n"
        f"{body}"
        "\\end{table}\n"
    )


# --- Usage ---------------------------------------------------------------
long_df = _melt_combined(combined)
for cohort in ['A', 'B']:
    wide = _build_cohort_wide(long_df, cohort)
    print(f"\n% === Cohort {cohort}: MDC stats for all composite scores ({now()}) ===")
    print(cohort_to_latex(wide, cohort))
```

```python
import numpy as np
import pandas as pd

TASK_ORDER = ['journaling', 'cookieTheft', 'picnicScene', 'avg_speech_prediction']
TASK_DISPLAY = {
    'cookieTheft': 'Cookie Theft',
    'picnicScene': 'Picnic Scene',
    'journaling': 'Journaling',
    'avg_speech_prediction': 'Avg. Speech Pred.',
}
DOMAIN_ORDER = ['language', 'memory', 'executive_function',  'speed']
DOMAIN_DISPLAY = {
    'executive_function': 'Executive Function',
    'language': 'Language',
    'memory': 'Memory',
    'speed': 'Speed',
}

N_BOOTSTRAP_SAMPLES = 10000  # adjust to match your n_bootsrap_samples
ALPHA = 0.05                 # threshold for flagging the non-inferiority p value
P_MARK = r'$^{\dagger}$'     # marker appended when p < ALPHA (non-inferiority established)

# Rows of `combined` that carry the numbers we need, e.g.
#   'journaling_bootstrap_ci_low'  or  'journaling_noninferiority_p_value'
_ROW_PATTERN = r'^(?P<task>.+?)_(?P<block>bootstrap|noninferiority)_(?P<metric>.+)$'

_FLOAT_COLS = [
    'mean_diff', 'ci_low', 'ci_high', 'prop_speech_worse',
    'ni_margin', 'ni_ci_low', 'ni_ci_high', 'ni_p_value',
]


def _melt_combined(combined):
    """Reshape `combined` from
        index = '<task>_bootstrap_<metric>' / '<task>_noninferiority_<metric>',
        columns = (domain, cohort)
    to a long dataframe with columns
        cohort, domain, task, mean_diff, ci_low, ci_high, excludes_zero,
        prop_speech_worse, ni_margin, ni_ci_low, ni_ci_high, ni_p_value.

    Non-inferiority metrics are prefixed with 'ni_' so they cannot collide with
    the bootstrap ones; every other row of `combined` is ignored.
    """
    # Split the row index into (task, block, metric); drop rows we do not model.
    parsed = combined.index.to_series().str.extract(_ROW_PATTERN)
    keep = parsed['task'].notna().to_numpy()

    tmp = combined.loc[keep].copy()
    task = parsed.loc[keep, 'task'].to_numpy()
    block = parsed.loc[keep, 'block'].to_numpy()
    metric = parsed.loc[keep, 'metric'].to_numpy()
    metric = np.where(block == 'noninferiority', np.char.add('ni_', metric.astype(str)), metric)
    tmp.index = pd.MultiIndex.from_arrays([task, metric], names=['task', 'metric'])

    # Long form: one row per (task, metric, domain, cohort).
    long = tmp.stack(level=[0, 1], future_stack=True).rename('value').reset_index()
    long.columns = ['task', 'metric', 'domain', 'cohort', 'value']

    # Pivot metrics into columns.
    wide = long.pivot_table(
        index=['cohort', 'domain', 'task'],
        columns='metric',
        values='value',
        aggfunc='first',
    ).reset_index()
    wide.columns.name = None

    # Types. Metrics that are absent (e.g. no non-inferiority test run) become NaN.
    for c in ['excludes_zero'] + _FLOAT_COLS:
        if c not in wide.columns:
            wide[c] = np.nan
    wide['excludes_zero'] = (
        wide['excludes_zero'].replace("True", 1).replace("False", 0).fillna(0).astype(bool)
    )
    for c in _FLOAT_COLS:
        wide[c] = pd.to_numeric(wide[c], errors='coerce')
    return wide


def _fmt_ci(row):
    s = f"{row['ci_low']:.2f} -- {row['ci_high']:.2f}"
    if row['excludes_zero']:
        s += '*'
    return s


def _fmt_prop(p):
    return f"{p:.1%}".replace('%', r'\%')


def _fmt_p(p, alpha=ALPHA):
    """Format the one-sided non-inferiority p value; flag it when p < alpha."""
    if pd.isna(p):
        return '--'
    s = r'$<0.001$' if p < 0.001 else f'{p:.3f}'
    if p < alpha:
        s += P_MARK
    return s


def _build_cohort_wide(long_df, cohort_label):
    """Build the (task x (domain, metric)) wide table for a single cohort."""
    sub = long_df[
        (long_df['cohort'] == f'Cohort {cohort_label}')
        & (long_df['task'].isin(TASK_ORDER))
    ].copy()
    sub['ci_str'] = sub.apply(_fmt_ci, axis=1)
    sub['prop_str'] = sub['prop_speech_worse'].apply(_fmt_prop)
    sub['p_str'] = sub['ni_p_value'].apply(_fmt_p)

    pieces = []
    for domain in DOMAIN_ORDER:
        d = (
            sub[sub['domain'] == domain]
            .set_index('task')[['ci_str', 'prop_str', 'p_str']]
        )
        d.columns = pd.MultiIndex.from_tuples([
            (DOMAIN_DISPLAY[domain], r'95\% CI'),
            (DOMAIN_DISPLAY[domain], r'\% larger'),
            (DOMAIN_DISPLAY[domain], r'$p_{\mathrm{NI}}$'),
        ])
        pieces.append(d)

    wide = pd.concat(pieces, axis=1).loc[TASK_ORDER]
    wide.index = [TASK_DISPLAY[t] for t in wide.index]
    wide.index.name = None
    return wide


def _get_margin(long_df, cohort_label):
    """The non-inferiority margin used for this cohort (None if not unique)."""
    m = long_df.loc[long_df['cohort'] == f'Cohort {cohort_label}', 'ni_margin'].dropna().unique()
    return float(m[0]) if len(m) == 1 else None


def cohort_to_latex(wide, cohort_label, n_bootstrap=N_BOOTSTRAP_SAMPLES, margin=None, alpha=ALPHA):
    column_format = 'l' + 'c' * wide.shape[1]
    body = wide.to_latex(
        column_format=column_format,
        multicolumn=True,
        multicolumn_format='c',
        escape=False,
    )
    margin_str = f"{margin:g}" if margin is not None else "the pre-specified margin"
    caption = (
        f"Bootstrap test results assessing whether the speech-based MDC is significantly larger "
        f"than the cognitive assessment's MDC on Cohort {cohort_label}. For each task and composite "
        f"cognitive score, we report the 95\\% CI of the difference (speech MDC minus cognitive MDC) "
        f"across {n_bootstrap} bootstrap replicates, the proportion of replicates where the speech "
        f"MDC is larger (less sensitive) than the cognitive MDC (\\% larger), and the one-sided "
        f"bootstrap p value of the non-inferiority test ($p_{{\\mathrm{{NI}}}}$) against a margin of "
        f"{margin_str}. An asterisk (*) on the CI indicates that the CI excludes zero, i.e. the speech "
        f"MDC is significantly larger than the cognitive MDC. A dagger ($\\dagger$) on "
        f"$p_{{\\mathrm{{NI}}}}$ indicates $p<{alpha:g}$, i.e. the speech MDC is significantly "
        f"non-inferior to the cognitive MDC at the {margin_str} margin."
    )
    label = f"tab:mdc_bootstrap_cohort{cohort_label}"
    return (
        "\\begin{table}\n"
        f"\\caption{{{caption}}}\n"
        f"\\label{{{label}}}\n"
        f"{body}"
        "\\end{table}\n"
    )


# --- Usage ---------------------------------------------------------------
long_df = _melt_combined(combined)
for cohort in ['A', 'B']:
    wide = _build_cohort_wide(long_df, cohort)
    print(f"\n% === Cohort {cohort}: MDC stats for all composite scores ({now()}) ===")
    print(cohort_to_latex(wide, cohort, margin=_get_margin(long_df, cohort)))
```
